# Agentic Travel Data Platform

Design a backend platform providing unified, natural-language access to data from multiple travel agencies.

## Requirements

1. Ingest data from data layer, SQL/PostgreSQL/NoSQL/CSV and multiple, evolving travel-agency sources.
2. Federated access to sources — no requirement to merge everything into one database.
3. Roles (Manager, Revenue Manager, Basic User) whose required data spans multiple sources.
4. Business Services Layer (Booking, Revenue, Agency, Travel), architected to scale to ~50 services.
5. Non-functional: scalability, reliability, flexibility, security, fault tolerance.
6. Smart Data Accessibility Layer: natural-language requests, e.g. *"Show me revenue about Pace for the last week and prepare a travel plan based on 20% of my revenue."*
7. Understand intent, identify required operations, route to the relevant service(s), support multi-service requests, and keep session context.
8. Answer with XAI (explainable AI) grounded in the underlying service response(s).
9. For derived/computed requests: provide result XAI, evidence/provenance, explanation of calculations/decisions, and confidence level.
10. Ask for clarification on ambiguous or conflicting requests and continue in conversational mode.
11. Do not re-ask for details already clarified earlier in the conversation.

## Examples for Requests

1. *"Show me revenue about Pace"* — return data from the single Revenue service.
2. *"Show me revenue about Pace and my Booking for the last week"* — return collected data from the Revenue and Booking services.
3. *"Show me revenue about Pace for the last week and prepare a travel plan based on 20% of my revenue."* — return data from Revenue + do text analysis to prepare a "travel plan based on 20% of my revenue".

## Constraints

1. Backend and APIs only — no UI.
2. No REST.
3. Streaming responses — user sees the answer line by line.
4. Distributed tracing and observability.
5. Authentication and authorization.

## Interview Goal

Design the high-level architecture and explain:

1. Main components and their responsibilities
2. Communication between components
3. Data federation approach
4. Service-to-service communication
5. Intent detection and routing
6. Handling of failures and partial results
7. Scalability to ~50 services
8. Security and authorization
9. Streaming for the text responses (user should see text response line by line)
10. Observability and tracing
